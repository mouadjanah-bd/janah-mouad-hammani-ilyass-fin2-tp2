```python
import pandas as pd

file_path = '/content/drive/MyDrive/janah mouad et hammani ilyass fin 2 tp 2/extportal.xls'

# Lecture du fichier TSV en sautant les 4 premières lignes pour garder les bons en-têtes de colonnes
try:
    df = pd.read_csv(file_path, sep='\t', skiprows=4)
    print("Le tableau a été chargé avec succès avec les bons en-têtes !")
    display(df.head())
except Exception as e:
    print(f"Erreur lors de la lecture : {e}")
```

```python
import sqlite3
import pandas as pd

# Base locale temporaire dans l'environnement Colab
conn = sqlite3.connect('tp_sql.db')

# Importer les données dans la table SQLite "donnees"
df.to_sql('donnees', conn, if_exists='replace', index=False)

# Première requête SQL
requete = '''
SELECT *
FROM donnees
LIMIT 10;
'''
resultat = pd.read_sql_query(requete, conn)
display(resultat)

# Vérifier le nombre de lignes importées
total = pd.read_sql_query(
    'SELECT COUNT(*) AS total FROM donnees;', conn
)
display(total)

conn.close()
```

```python
import sqlite3
import pandas as pd

# Connexion à la base de données
conn = sqlite3.connect('tp_sql.db')

# --- Requête 1 : Calcul de la somme totale et de la moyenne de la colonne 'Total Holdings'
print("=== Requête 1 : Somme globale et Moyenne de 'Total Holdings' ===")
requete_calcul = '''
SELECT
    SUM("Total Holdings") AS Somme_Totale_Holdings,
    AVG("Total Holdings") AS Moyenne_Holdings
FROM donnees;
'''
res_calcul = pd.read_sql_query(requete_calcul, conn)
display(res_calcul)

# --- Requête 2 : Filtrer les dates récentes et afficher la part (calculée) du 'Securities' dans le 'Total Holdings'
print("\n=== Requête 2 : Calcul du pourcentage de 'Securities' par rapport au total pour les années récentes ===")
requete_ratio = '''
SELECT
    Date,
    Securities,
    "Total Holdings",
    (CAST(REPLACE(Securities, ',', '') AS REAL) / CAST(REPLACE("Total Holdings", ',', '') AS REAL)) * 100 AS Pourcentage_Securities
FROM donnees
WHERE Date LIKE '%202%'
LIMIT 5;
'''
res_ratio = pd.read_sql_query(requete_ratio, conn)
display(res_ratio)

# --- Requête 3 : Trier les données pour trouver les 5 dates avec le plus grand 'Nbr 1 Account'
print("\n=== Requête 3 : Top 5 des dates avec le montant le plus élevé dans 'Nbr 1 Account' ===")
requete_top = '''
SELECT
    Date,
    "Nbr 1 Account"
FROM donnees
ORDER BY CAST(REPLACE("Nbr 1 Account", ',', '') AS REAL) DESC
LIMIT 5;
'''
res_top = pd.read_sql_query(requete_top, conn)
display(res_top)

conn.close()
```
